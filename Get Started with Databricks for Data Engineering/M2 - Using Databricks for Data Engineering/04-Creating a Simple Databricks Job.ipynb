# 간단한 Databricks 작업 만들기

Databricks Jobs(Lakeflow 작업)는 Databricks에서 모든 처리 작업을 예약하고 오케스트레이션할 수 있는 도구 모음을 제공합니다.

**목표:** 이전 데모에서 빌드한 Pipeline을 사용하여 작업에 두 개의 태스크를 만듭니다. 파이프라인는 데모를 위해 두 개의 Notebooks로 구분되었습니다.
- **DEWD00 - 04A-Task 1 - Setup - Bronze**
- **DEWD00 - 04B-Task 2 - Silver - Gold**


**참고:** 이 데이터 엔지니어링 작업에는 Lakeflow Spark Declarative Pipeline을 사용할 수도 있었지만, Spark Declarative Pipeline은 본 과정의 범위를 벗어납니다. SDP는 추가 작업과 함께 Lakeflow Job 내에서 스케줄링될 수 있습니다.


## 환경 준비 (Databricks Free Edition)

- **카탈로그** `db_catalog`, **스키마** `db_schema` 를 사용합니다.
- 스키마 `db_schema` 아래 **`myfiles`** 볼륨을 만들고, **`employees.csv`** 를 업로드합니다.
- (다른 노트북에서 안내한 것과 동일한 데이터 레이아웃을 사용합니다.)



```sql
%sql
USE CATALOG db_catalog;
USE SCHEMA db_schema;
```


## 1. Lakeflow 작업 구성 생성

아래 셀을 실행하면 작업 UI에 넣을 작업 이름 예시가 출력됩니다. (실제로는 원하는 이름을 사용해도 됩니다.)


```sql
%python
import uuid
job_suggested_name = "de_m2_lakeflow_job_" + uuid.uuid4().hex[:8]
print("Jobs UI에서 작업 이름 예시:", job_suggested_name)
print("노트북 태스크 경로: 워크스페이스에서 DEWD00 - 04A / 04B 노트북의 전체 경로를 지정하세요.")
```


## 2. 노트북 태스크로 작업 구성

Jobs UI를 사용하여 여러 태스크로 워크로드를 오케스트레이션하는 경우 항상 단일 태스크로 Job을 생성하는 것으로 시작하고 필요한 경우 더 추가할 수 있습니다.

위의 Notebooks(**DEWD00 - 04A-Task 1 - Setup - Bronze** 및 **DEWD00 - 04B-Task 2 - Silver - Gold**)를 사용하여 두 개의 태스크가 있는 Lakeflow 작업을 생성하려면 다음을 완료합니다.

1. 사이드바에서 **작업 및 파이프라인** 버튼을 마우스 오른쪽 단추로 클릭하고 새 탭에서 링크를 엽니다. 이렇게 하면 필요에 따라 이러한 지침을 참조할 수 있습니다.

2. **작업 및 파이프라인** 탭에 있는지 확인합니다.

3. 오른쪽에서 **만들기 -> 작업**을 선택합니다.

4. 화면 왼쪽 상단에 위에 제공된 **작업 이름**을 입력 하여 Lakeflow 작업의 이름을 추가합니다.

5. **첫 번째 태스크 추가 (Add your first task)** 아래에서 **노트북 (Notebook)** 을 선택하세요. **노트북 (Notebook)** 이 목록에 없으면 **+ 다른 태스크 유형 추가 (+ Add another task type)** 를 클릭하고 옵션에서 **노트북 (Notebook)** 을 선택하세요.

6. 아래 지침에 따라 태스크를 설정하세요. (edited)


### 태스크 1 만들기
| 설정 | 지침 |

| 소스 | **워크스페이스**가 선택되었는지 확인하세요. |
| 경로 | 탐색기를 사용하여 **DEWD00 - 04A-Task 1 - Setup - Bronze** 노트북을 지정하십시오. 위의 경로를 참고하여 노트북을 찾으십시오. |
| **Compute** | 드롭다운 메뉴에서 **서버리스** 클러스터를 선택하세요.<br>(본 과정의 모든 작업에는 서버리스 클러스터를 사용합니다. 필요 시 과정 외부에 다른 클러스터를 지정할 수 있습니다.)<br></br>참고: 범용 클러스터를 선택할 경우, 이 작업이 범용 Compute으로 청구된다는 경고가 표시될 수 있습니다. 프로덕션 작업은 항상 작업 부하에 적합한 크기의 새 작업 클러스터에 대해 예약해야 합니다. 이는 훨씬 낮은 요금으로 청구되기 때문입니다. |
| 환경 및 라이브러리 | **기본값**이 선택되었는지 확인하세요. |
| 만들기 | **태스크 만들기** 단추를 선택하여 태스크를 만듭니다. |

### 태스크 2 만들기

####더 나은 성능을 위해 작업 세부 정보에서 성능 최적화 모드를 활성화하십시오. 그렇지 않으면 실행 시작에 6~8분이 소요될 수 있습니다.



## 3. 예약 옵션 살펴보기
다음 단계를 완료하여 스케줄링 옵션을 탐색하십시오.

1. 작업 UI의 오른쪽에서 **일정 및 트리거** 섹션을 찾습니다.

2. **트리거 추가** 단추를 선택하여 스케줄링 옵션을 탐색합니다.

3. **트리거 유형**을 **없음 (수동)** 에서 **스케줄링**으로 변경 하면 cron 스케줄링 UI가 나타납니다.

   - 이 UI는 LakeFlow 작업의 크론 스케줄링을 설정하기 위한 광범위한 옵션을 제공합니다. UI로 구성된 설정은 cron 구문으로 출력할 수도 있으며, UI가 필요할 때가 아니라 사용자 지정 구성을 사용할 수 없는 경우 편집할 수 있습니다.

4. **취소**를 선택하여 작업 세부 정보로 돌아갑니다.

## 4. 작업 실행
**작업 세부** 정보 위의 **지금 실행**을 선택하여 작업을 실행합니다.

## 5. 작업 실행 검토

작업 실행을 검토하려면 다음을 수행합니다.

1. 작업 세부 정보 페이지에서 화면 왼쪽 위에 있는 **실행** 탭을 선택합니다 (현재 **작업** 탭에 있어야 함).

2. 작업을 찾으세요.

3. **시작 시간** 열 아래의 타임스탬프 필드를 클릭하여 출력 세부 정보를 엽니다.

    - **작업이 아직 실행** 중인 경우, 오른쪽 패널에 노트북의 활성 **상태**가 상태 또는 `Pending`와 함께 `Running` 표시됩니다.

    - **작업이 완료되면** 오른쪽 패널에 노트북의 전체 실행이 상태 `Succeeded` 또는 `Failed`로 표시됩니다.
