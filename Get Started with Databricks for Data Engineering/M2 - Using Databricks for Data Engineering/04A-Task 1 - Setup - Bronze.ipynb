### BRONZE
**목표:** **myfiles** 볼륨의 모든 CSV 파일을 사용하여 테이블을 만듭니다.

```sql
-- 빈 테이블 및 열 만들기
CREATE TABLE IF NOT EXISTS current_employees_bronze_job (
  ID INT,
  FirstName STRING,
  Country STRING,
  Role STRING
  );
```

```sql
%python

# 브론즈 원시 수집 테이블을 만들고 행의 파일 이름을 포함합니다.
spark.sql(f'''
  COPY INTO current_employees_bronze_job
  FROM '/Volumes/db_catalog/db_schema/myfiles/'
  FILEFORMAT = CSV
  FORMAT_OPTIONS (
    'header' = 'true', 
    'inferSchema' = 'true'
)
''').display()
```
