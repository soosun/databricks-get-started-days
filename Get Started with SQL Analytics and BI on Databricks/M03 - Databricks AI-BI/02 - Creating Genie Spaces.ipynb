## 02 - Genie Spaces 생성

이번 수업에서는 Databricks AI/BI Genie와 기존 대시보드와 데이터 세트, 그리고 workspace에 연결된 새로운 데이터 세트 조합을 기반으로 만들 수 있는 데이터 탐색 공간을 살펴볼 것입니다. 

이 수업은 **db_catalog** 카탈로그의 **db_schema** 스키마에서 다음 자료들을 사용합니다:
* **sales** 테이블
* **customers** 테이블

또한 다음 사항을 생성해야 합니다:
  * _수업: Databricks에서 대시보드 만들기 (소매 대시보드)에서 만든 대시보드_

**참고:** **Databricks Free Edition**(M01/M02와 동일)에서는 카탈로그 **db_catalog**, 스키마 **db_schema**, 볼륨 **sqlfiles**에 CSV를 올려 **sales**, **customers**, **sales_orders** 테이블을 준비해 두세요. 이미 해당 테이블이 있으면 그대로 사용하면 됩니다.

### 1부: Genie Space 만들기
데모의 이 부분에서는 주어진 UI 영역에서 직접 Genie Space를 생성하는 것부터 시작할 것입니다. 아래 단계를 따라 Genie Space을 만드세요. 
1. 플랫폼 왼쪽 내비게이션에서 **Genie**로 이동하세요. 
2. 오른쪽 상단의 **+ 새 항목 (+ New)** 을 클릭하세요.
3. 새로운 팝업이 나타나 데이터를 연결하라는 안내를 받을 것입니다. 이 팝업에서 **모든 (All)** 을 선택하여 테이블을 찾으세요:
      - 카탈로그: db_catalog
      - 스키마: db_schema
      - 테이블: customers
4. 데이터 선택이 끝나면 하단의 **생성 (Create)** 을 클릭하세요.

이제 왼쪽에 채팅 환경, 오른쪽에 설정과 세부 정보가 있는 Genie Space UI가 보입니다. 상단의 **구성 (Configure)** 버튼을 선택한 후, **설정 (Settings)** 를 클릭하세요. (기본적으로 구성은 **데이터 (Data)** 로 열립니다.) 여기에서 다음 정보를 편집할 수 있습니다:
1. **제목 (Title):** 기본 소매 세부사항
2. **설명 (Description):** "이 공간은 고객의 정보를 쿼리할 수 있는 공간을 제공하기 위해 설계되었습니다 데이터세트."
3. **기본 웨어하우스 (Default warehouse):** shared_warehouse
4. **샘플 질문 (Sample Questions):** "캘리포니아에 고객이 몇 명인가요?" (질문을 추가하려면 **+ 추가 (+ Add)** 버튼을 클릭하세요.)
5. 편집 내용을 확인하려면 하단의 **저장 (Save)** 을 클릭하세요.

Genie space의 화면 영역은 두 면으로 나뉘어 있는데, 채팅창은 왼쪽에, 구성과 설정은 오른쪽에 있습니다. 오른쪽 상단의 버튼으로 다음 영역 중에서 선택할 수 있습니다:

* **새 채팅 (New chat)**: Genie로 새로운 스레드 대화를 만들 수 있습니다. 최종 사용자에게 Genie space을 게시하면(이들은 아마도 "Can View" 또는 "Can Run" 접근 권한이 있을 가능성이 큽니다), 이 공간은 그들이 접근할 수 있는 몇 안 되는 Genie 영역 중 하나입니다.
* **역사 (History)**: Genie와 나눈 개별 채팅 스레드를 검토할 수 있게 해줍니다. 최종 사용자에게 Genie space을 게시하면(아마도 "Can View" 또는 "Can Run" 접근 권한이 있을 가능성이 큽니다), 이 공간이 그들이 접근할 수 있는 또 다른 Genie space입니다.
* **구성 (Configure)**: Genie space 설정 편집 화면으로 돌아가게 해줍니다. 공간 생성 시 본 화면과 비슷합니다.
  구성 내에서는 다음과 같은 기능이 있습니다:
  * **데이터 (Data)**: 공간과 연관된 데이터 테이블을 검토하고 편집할 수 있습니다.
  * **지침 (Instructions)**: 사용자가 질문했을 때 Genie가 어떻게 행동할지에 대해 자연어로 일반적인 지침을 제공할 수 있습니다.
  * **설정 (Settings)**: 제목, Default Warehouse, 설명, 샘플 질문을 지정하여 공간을 구성할 수 있습니다.
* **벤치마크 (Benchmarks)**: 중요한 사용자 질문에 대해 좋은 답변을 계속 제공할 수 있도록 반복적으로 실행하는 질문 모음을 정의할 수 있습니다.
* **모니터링 (Monitoring)**: 어떤 질문이 있었는지, 누가 질문했는지, 그리고 사용자가 어떻게 평가했는지 검토할 수 있게 해줍니다. 
* **공유 (Share)**: 공유 권한을 설정하고 Genie space을 최종 사용자와 공유할 수 있게 해줍니다. 

또한, kebab menu **(⋮)** 아래에서 Genie space을 복제하거나 인증을 할당하거나, 이 메뉴에서 휴지통으로 이동하여 삭제하는 옵션을 찾을 수 있습니다.

### 2부: 대시보드에서 Genie Space 생성
또는 대시보드에서 직접 Genie Space을 만들 수도 있습니다. 

1. **대시보드 (Dashboards)** 로 이동하여 **Lesson - Creating a Dashboard in Databricks**(소매 대시보드)에서 만든 대시보드를 선택하세요.

2. 대시보드를 **초안 편집 (Edit Draft)** 뷰로 전환하세요.

3. 오른쪽 상단의 **(⋮)** Kebab menu 아이콘을 클릭하고 **설정 (Settings)** 옵션을 선택하세요.

4. **설정 (Settings)** 사이드바 창에서 **일반 (General)** 탭을 선택하세요.

5. **Genie** 토글 버튼을 클릭해 **Genie**를 대시보드에 대해 활성화하세요.

    - 선택할 수 있는 선택권이 주어집니다 
        - 자동 생성 Genie Space
        - 기존 Genie Space 링크
    - 이 연습에서는 **Genie Space 자동 생성 (Auto-generate Genie space)** 을 선택하세요.

6. **게시 (Publish)** 를 클릭하여 출판 대화상자를 열어보세요. **데이터 공유 권한 (Share data permission)** 옵션은 기본값으로 남겨두고, 대화 오른쪽 하단의 **게시 (Publish)** 를 클릭하세요.

7. **공유 (Sharing)** 대화 상자를 닫고 대시보드의 **게시된 버전 보기 (View Published)** 로 이동합니다. 

8. 하단에 내장된 **Ask Genie** 버튼을 클릭하세요. 이렇게 하면 dashboard 위에 팝업 채팅창이 열립니다. Kebab menu 버튼을 사용해 Genie 채팅을 화면 옆에 도킹하는 설정에 접근할 수 있습니다.

9. 채팅창에 다음 질문을 하세요.

    _어떤 테이블이 있고 어떻게 연결되어 있나요? 간단히 요약해 주세요._
10. Genie가 제공한 답변을 검토하세요.
