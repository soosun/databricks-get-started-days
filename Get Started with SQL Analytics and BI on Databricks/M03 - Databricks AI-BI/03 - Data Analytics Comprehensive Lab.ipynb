## 03 - 데이터 분석 종합 연구실

가상의 소매 조직의 데이터 분석가로서, 경영진은 04-AI/BI Dashboards에서 생성된 대시보드에 추가 데이터 세트를 추가해 달라고 요청했습니다. 또한 몇 가지 더 많은 시각화를 추가해 달라고 하는데, 이는 몇 가지 추가 쿼리가 필요할 것입니다. 
마지막으로, 결과된 대시보드에서 Genie Space를 만들어 부재 중에도 추가 질문에 답할 수 있도록 하세요. 

따라야 할 단계들은 다음과 같습니다:
- 데이터세트를 sales_orders 찾아 탐색하세요.
- 대시보드에 새 데이터세트를 추가하세요.
- 데이터세트 목록에 새로운 쿼리를 추가하세요.
- 대시보드에 새로운 시각화를 추가하세요.
- 필터를 조정하여 새로운 변화를 반영하세요.
- 대시보드를 게시하세요.
- 대시보드에서 Genie Space을 만드세요. 

이 수업은 **db_catalog** 카탈로그의 **db_schema** 스키마에서 다음 자료들을 사용합니다:
* **sales_orders** 테이블

또한 다음 사항을 생성해야 합니다:
  * _수업: Databricks에서 대시보드 만들기 (소매 대시보드)에서 만든 대시보드_

**참고:** **Databricks Free Edition**(M01/M02와 동일)에서는 카탈로그 **db_catalog**, 스키마 **db_schema**, 볼륨 **sqlfiles**에 CSV를 올려 **sales**, **customers**, **sales_orders** 테이블을 준비해 두세요. 이미 해당 테이블이 있으면 그대로 사용하면 됩니다.

### 1부: 대시보드에 새로운 데이터셋 추가하기

이제 소매 조직 대시보드를 위해 추가 테이블을 데이터 세트로 추가하여 더 많은 데이터를 준비할 차례입니다.

1. 사이드바 메뉴에서 **대시보드 (Dashboards)** 를 클릭하면 대시보드 목록 페이지로 이동합니다. 기본적으로 당신이 소유한 대시보드 목록이 표시됩니다.

2. **소매 대시보드 (Retail Dashboard)** 를 클릭하면 수업에서 생성된 대시보드를 엽니다(Databricks에서 대시보드 생성). 대시보드의 마지막 출판본이 열립니다.

3. 대시보드 상단의 대시보드 이름 옆에 있는 컨텍스트 메뉴 아이콘을 사용해 대시보드의 **초안 편집 (Edit Draft)** 버전을 확인하세요.

4. **데이터 (Data)** 탭을 클릭하세요.

5. **+ 데이터 소스 추가 (+ Add data source)** 를 클릭하여 새 테이블을 추가하세요.

6. **db_catalog** → **db_schema**에서 **sales_orders** 테이블을 찾아 데이터세트로 추가하세요.

### 3부: SQL 쿼리로 데이터세트 추가하기

다음 두 차트는 서로 다른 두 테이블의 데이터가 필요합니다. <b>데이터 (Data)</b> 탭의 쿼리 편집기에서 `JOIN`을 사용하여 필요한 모든 값을 포함하는 데이터 세트를 생성할 수 있습니다. 데이터 세트는 공통 테이블 표현식과 조인을 사용하여 생성할 수 있습니다.

#### 데이터세트 1: 충성도 세그먼트별 매출

이 쿼리는 고객과 판매 테이블의 데이터를 조인하는 데이터세트를 생성합니다. 그것은 고객의 지출 금액과 로열티 프로그램 상태를 연결합니다.

이 데이터세트를 만들기 위해 다음 단계를 완료하세요:

1. <p style="color: #0873FF"> <b>데이터 (Data)</b> 탭을 클릭하세요. </p>
2. <p style="color: #0873FF"> <b>SQL에서 만들기 (Create from SQL)</b>를 클릭하세요. </p>
3. <p style="color: #0873FF"> 새 데이터 세트의 제목을 두 번 클릭하고 기본 이름인 <b>제목 없는 데이터 세트 (Untitled dataset)</b>를 <b>로열티 세그먼트별 매출 (Sales by loyalty segment)</b>로 변경하세요.</p> 
4. <p style="color: #0873FF"> 다음 쿼리를 편집기에 붙여넣으세요: </p>

```
SELECT
      product_category,
      loyalty_segment,
      total_price
    FROM
      db_catalog.db_schema.sales AS sales
      JOIN db_catalog.db_schema.customers AS customers
      ON sales.customer_id = customers.customer_id
```

5. <p style="color: #0873FF"> 쿼리에 **db_catalog.db_schema** 아래 올바른 테이블이 지정되었는지 확인하세요. </p>

6. <p style="color: #0873FF"> 쿼리를 실행하세요. </p>

#### 데이터세트 2: 일별 판매 및 주문 </a>

이 쿼리는 구매가 이루어진 월의 요일, 그날의 총 세일즈, 그리고 해당 그룹의 주문 수를 보여주는 데이터세트를 생성합니다. 데이터는 `sales` 및 `sales_order` 테이블에서 가져옵니다.

다음 데이터세트를 추가하려면 다음 단계를 완료하세요:

1. <p style="color: #0873FF"> <b>SQL에서 만들기 (Create from SQL)</b>를 클릭하세요. </p>
2. <p style="color: #0873FF"> 새 데이터 세트의 제목을 두 번 클릭하고 기본 이름인 <b>제목 없는 데이터 세트 (Untitled dataset)</b>를 <b>일별 판매 및 주문 현황 (Sales and orders by day)</b>로 변경하세요.</p>
3. <p style="color: #0873FF"> 다음 쿼리를 편집기에 붙여넣으세요: </p>

```
WITH sales_data AS (
  SELECT
    date_format(order_date, "dd") AS day,
    SUM(total_price) AS total_sales
  FROM db_catalog.db_schema.sales
  GROUP BY day
),
orders_data AS (
  SELECT
    CASE 
      WHEN try_cast(sales_orders.order_datetime AS BIGINT) IS NOT NULL 
      THEN DAY(FROM_UNIXTIME(sales_orders.order_datetime))
      ELSE NULL 
    END as day,
    COUNT(order_number) AS total_orders
  FROM db_catalog.db_schema.sales_orders
  GROUP BY day
)
SELECT
  cast(s.day as INT),
  s.total_sales,
  o.total_orders
FROM sales_data s
JOIN orders_data o ON s.day = o.day
ORDER BY s.day;
```

4. <p style="color: #0873FF"> 쿼리에 **db_catalog.db_schema** 아래 올바른 테이블이 지정되었는지 확인하세요. </p>

5. <p style="color: #0873FF"> 쿼리를 실행하세요. </p>

### 4부: 추가 시각화 만들기

이제 그 새로운 데이터세트를 활용해 새로운 시각화를 만듭시다.

#### 히트맵 시각화

히트맵은 특정 사건의 발생 패턴을 이해하는 데 도움이 됩니다. 곧 생성하려는 히트맵은 고객이 할당한 충성도 세그먼트와 제품 카테고리별로 데이터를 나누어 보여줍니다. 차트의 색상은 각 카테고리에서 지출된 총 금액을 나타냅니다.

히트맵을 작성하려면 다음 단계를 완료하세요:

1. <p style="color: #0873FF"> <b>제목 없는 페이지 (Untitled page)</b>캔버스 탭을 클릭하세요. </p>
2. <p style="color: #0873FF"> 캔버스에 시각화 위젯을 추가하세요. </p>
3. <p style="color: #0873FF"> 다음 설정 설정을 적용하세요: </p>

    - <b>데이터셋 (dataset)</b>: Sales by loyalty segment
    - <b>시각화 (Visualization)</b>: 히트맵
    - <b>X 축</b>: product_category
    - <b>Y축</b>: loyalty_segment
    - <b>색상(Color)</b>: SUM(total_price)

4. <p style="color: #0873FF"> dashboard에서 위젯을 재조정하고 배열하세요. </p>
5. <p style="color: #0873FF"> 이 위젯의 기본 크기는 모든 제품 카테고리를 보여주기에는 너무 작습니다. 위젯의 가장자리 위에 마우스를 올리고 차트 크기를 조절하세요. 차트의 다른 영역 위에 마우스를 올려서 손 아이콘이 보일 때까지 기다리세요. 클릭하고 드래그하면 차트를 캔버스 위로 이동할 수 있습니다. </p>

좋습니다! 이 데이터에서 그것은 충성도 세그먼트 3 고객이 전체적으로 가장 많은 지출을 하는 것으로 보입니다. 제품 카테고리 <b>Reagate</b>와 <b>Zamaha</b>에서는 충성도 세그먼트가 확산되는 데 더 느리며 고객 지출과는 크게 관련이 없어 보입니다.

#### 이중 축 선 차트


이중 축 선 차트는 이중 축 막대 차트와 마찬가지로 서로 다른 척도에서 관련 양을 비교하는 데 유용할 수 있습니다. 곧 만들 차트는 세일즈와 주문 수에서 발생하는 수익을 추적합니다. 데이터는 매각이 이루어진 월의 일별로 그룹화되어 있습니다. 이는 한 달 내내 고객 지출 패턴을 추적하는 데 유용할 수 있습니다.

dual-axis line chart를 만들려면 다음을 작성하세요:

1. <p style="color: #0873FF"> 캔버스에 시각화 위젯을 추가하세요. </p>
2. <p style="color: #0873FF"> 다음 설정 설정을 적용하세요: </p>

    - <b>데이터셋 (Dataset)</b>: Sales and orders by day
    - <b>시각화 (Visualization)</b>: 선
    - <b>X 축:</b> 일 (<b>변환</b>을 <b>None</b>으로 설정)
    - <b>Y축</b>: SUM(total_sales)
    - <b>Y축</b>: SUM(total_orders)
    이렇게 하면 기본 차트가 생성되지만, 이중 축을 활성화해 각 척도의 수량을 표시해야 합니다.

3. <p style="color: #0873FF"> Y축 오른쪽의 케밥 아이콘을 클릭하세요. 그런 다음, <b>이중 축 활성화</b>를 켜기 위해 체크박스를 클릭하세요. </p>
4. <p style="color: #0873FF"> dashboard에서 위젯을 재조정하고 배열하세요. </p>

### 5부: 필터 편집 및 재게시

시각화 작업을 완료하셨으니, 시간이 더 된다면 대시보드에 추가 텍스트 설명과 차트의 대체 색상 표현으로 자유롭게 맞춤 설정할 수 있습니다. 완료 후에는 필터에 선택 필드를 추가해야 그것이 추가 차트에 영향을 미치도록 해야 합니다. 

##### 필터 편집
필터를 편집하려면 다음 단계를 사용하세요: 

1. 대시보드에서 필터를 선택하세요.
2. 설정 패널에서 필드 옆의 + 버튼을 선택하세요.
3. _sales by loyalty segment.product_category_ 필드를 추가합니다.

##### 재출판 대시보드
떠나기 전에, 추가 사용자가 완성된 대시보드를 뷰할 수 있도록 대시보드를 다시 게시하는 것을 잊지 마세요.

1. 수정된 대시보드의 공유 가능한 복사본을 생성하려면 **게시 (Publish)** 를 클릭하세요.
2. 대시보드에서 뷰 게시된 대시보드 버전 상단의 맥락 아이콘을 사용하세요.

### 보너스: 대시보드에서 Genie Space을 만들기 (작업 중)
대시보드에서 바로 Genie Space을 만드는 방법은 다음과 같은 단계를 사용하세요: 


1. 대시보드 항목은 **초안 편집 (Edit Draft)** 뷰로 전환하세요.

3. 오른쪽 상단의 **(⋮)** Kebab menu 아이콘을 클릭하고 **설정 (Settings)** 옵션을 선택하세요.

4. **설정 (Settings)** 사이드바 창에서 **일반 (General)** 탭을 선택하세요.

5. **Enable Genie** 토글 버튼을 클릭해 **Genie**를 대시보드에 대해 활성화하세요.

    - 선택할 수 있는 선택권이 주어집니다 
        - 자동 생성 Genie Space
        - 기존 Genie Space 링크
    - 이 연습에서는 **Genie Space 자동 생성 (Auto-generate Genie space)** 을 선택하세요.

6. **게시 (Publish)** 를 클릭하여 출판 대화상자를 열어보세요. **데이터 공유 권한 (Share data permission)** 옵션은 기본값으로 남기고, 대화 오른쪽 하단의 **게시 (Publish)** 를 클릭하면 대시보드를 다시 게시하세요.

7. **공유 (Sharing)** 대화 상자를 닫고 대시보드의 **게시된 버전 보기 (View Published)** 로 이동합니다. 

8. 하단에 내장된 **Ask Genie** 버튼을 클릭하세요. 이렇게 하면 dashboard 위에 팝업 채팅창이 열립니다. Kebab menu 버튼을 사용해 Genie 채팅을 화면 옆에 도킹하는 설정에 접근할 수 있습니다.

9. Genie Space을 탐색하고, 채팅창에 질문을 하며, 답변을 검토해 보세요.
