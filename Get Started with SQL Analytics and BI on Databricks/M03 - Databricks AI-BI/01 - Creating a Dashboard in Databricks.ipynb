## 01 - Databricks에서 대시보드 생성하기

Databricks AI/BI Dashboards는 향상된 시각화 라이브러리와 효율적인 구성 경험을 제공하여 데이터를 빠르게 공유 가능한 인사이트로 변환할 수 있도록 돕습니다. 

이번 수업에서는 새로운 대시보드를 만들고 테이블 데이터와 SQL 쿼리를 기반으로 대시보드에 데이터와 시각화를 추가합니다.

이 수업은 **db_catalog** 카탈로그의 **db_schema** 스키마에서 다음 자료들을 사용합니다:
* **sales** 테이블
* **customers** 테이블

이 테이블에는 일부 소매 판매 수치와 고객 주문 내역이 포함되어 있습니다. 우리는 이 자료들을 대시보드의 시각화 자료로 사용할 예정입니다.

**참고:** **Databricks Free Edition**(M01/M02와 동일)에서는 카탈로그 **db_catalog**, 스키마 **db_schema**, 볼륨 **sqlfiles**에 CSV를 올려 **sales**, **customers**, **sales_orders** 테이블을 준비해 두세요. 이미 해당 테이블이 있으면 그대로 사용하면 됩니다.

### 1부: 새 대시보드 생성
Databricks에서 새로운 대시보드를 만드는 것은 간단하고 명료합니다. 
* 측면 내비게이션 창에서 대시보드로 탐색하세요.
* **대시보드 생성 (Create dashboard)** 을 선택하세요. 
* 결과 화면 상단에서 대시보드 이름을 클릭하고 그것을 **Retail Dashboard**로 변경하세요.

이미 대시보드가 있다면 가져오는 옵션도 있습니다. 기존의 모든 대시보드는 이 플랫폼 구역에서 찾을 수 있습니다. 플랫폼 전역에 빠른 생성 기능도 많으며, 대시보드를 다른 하위 메뉴에서 생성하는 옵션 중 하나로 제공합니다.

### 2부: 데이터 추가

완전히 새로운 대시보드의 경우, 정보 표시 자체를 설계하기 전에 대시보드를 데이터와 연관시켜야 합니다. 대시보드 화면 상단에는 두 개의 탭, **데이터 (Data)** 와 **제목 없는 페이지 (Untitled page)** 가 있습니다. 

- **데이터:** **데이터** 탭에서는 대시보드에서 사용할 데이터 세트를 정의할 수 있습니다. 데이터 세트는 UI 또는 API를 통해 대시보드를 공유, 가져오기 또는 내보낼 때 함께 번들로 제공됩니다.

- **제목 없는 페이지:** 사용자가 시각화를 만들고 대시보드를 구성할 수 있게 하는 dashboard 캔버스입니다. 캔버스 위의 각 아이템은 위젯이라고 불립니다. 위젯은 세 가지 유형으로 나뉩니다: 시각화, 텍스트 박스, 필터 등이 있습니다.

**제목 없는 페이지 (Untitled page)** 탭에서 시작해서 제공되는 유용한 개요를 볼 수 있습니다. **데이터 (Data)** 탭을 선택하여 시작하세요. 

**데이터 (Data)** 탭 화면 왼쪽에는 세 개의 작은 아이콘이 있습니다: **데이터셋 (Datasets)**, **카탈로그 (Catalog)**, 그리고 **어시스턴트 (Assistant)**. 
* **데이터셋 (Datasets)** (목록을 나타내는 아이콘)은 대시보드에 사용된 모든 데이터셋 목록을 보여줍니다. 여기서 시작합니다. 현재는 비어 있습니다.
* **카탈로그 (Catalog)**(표준 Databricks 카탈로그 아이콘, 세 개의 작은 도형)은 workspace에서 접근 가능한 카탈로그, 스키마, 테이블을 탐색하고 데이터셋으로 사용할 테이블이나 물질화된 뷰를 선택할 수 있게 해줍니다.
* **어시스턴트 (Assistant)** (표준 Databricks Assistant 아이콘, 작은 네 갈래 별)는 자연어로 질문을 할 수 있는 AI 기반 인터페이스를 제공하며, 플랫폼에 물건을 발견하거나 인사이트 또는 쿼리 작성 지원을 받을 수 있습니다. 

다음 단계들이 이 예제 대시보드의 테이블을 추가하는 과정을 안내합니다.

1. **데이터셋 목록 (Datasets list)** 아이콘을 선택한 상태에서 **+ 데이터 소스 추가 (+ Add data source)** 버튼을 클릭하세요. </p>
2. 표시된 팝업에서 `db_catalog.db_schema.sales`를 검색하세요. </p>
3. **sales**를 클릭해 그것을 데이터세트로 추가한 후 **확인 (Confirm)** 버튼을 선택하세요. 참고 그것이 당신의 데이터세트 목록에 나타납니다.
</p>
4. 이 단계를 반복해서 `customers` 테이블를 추가하십시오. </p>

참고 각 테이블이 쿼리 편집 패널에서 자동으로 채워진 `SELECT *` 문으로 목록에 추가됩니다. SQL 쿼리를 수정해서 데이터세트를 바꿀 수 있습니다.

### 3부: 시각화 추가
#### 시각화 1: 카운터

대시보드에 추가할 첫 번째 시각화는 현재 매출을 300만 달러의 매출 목표와 비교하여 표시하는 카운터 시각화입니다.

1. **데이터 (Data)** 탭에서 **+ SQL에서 생성 (+ Create from SQL)** 옵션을 선택하세요. </p>
2. 다음 쿼리를 query 편집 공간에 입력하세요:
`SELECT sum(total_price) AS Total_Sales, 3000000 AS Sales_Goal FROM db_catalog.db_schema.sales;` </p>
3. 쿼리를 실행하려면 **실행 (Run)** 를 클릭하세요. </p>
4. **데이터셋 (Datasets)** 목록에서 쿼리를 우클릭하여 **이름 바꾸기 (Rename)** 를 선택하거나 Kebab menu을 사용해 쿼리를 **Count Total Sales**로 이름으로 바꾸세요. </p>
5. **제목 없는 페이지 (Untitled page)** 캔버스 탭로 돌아가기. </p>
6. 화면 하단에는 객체 이동, 시각화 추가, 텍스트 박스 추가, 필터 추가를 위한 툴바가 있습니다. **시각화 추가 (Add a visualization)** 선택하세요. </p>
7. 커서를 화면 어디든 이동시키고 캔버스에 시각화를 추가하려면 클릭하세요. </p>
8. 오른쪽의 **구성 패널 (Configuration Panel)** 에서 다음 설정을 선택하세요:
    - **제목 (Title):** 확인됨
      - 시각화에서 **위젯 제목 (Widget Title)** 을 클릭하세요.
      - **Sales Goal**로 변경하십시오.
    - **데이터셋 (dataset):** Count Total Sales
    - **시각화 (Visualization):** 카운터
    - **가치 (Value):** Total_Sales
    - **비교 (Comparison):** Sales_Goal

9. 이제 **Total_Sales**를 클릭하고 결과된 드랍다운에서 **형식 (Format)** 를 선택하세요. 그다음 **커스텀 (Custom)** 을 클릭하세요.  다음과 같은 조정을 하세요:
    - **유형 (Type):** 통화 ($)
    - **약어 (Abbreviation):** 없음

10. 스타일 섹션에서 **조건부 스타일 (Conditional Style)** 옆의 **+** 를 클릭하세요. 다음 설정으로 구성하십시오:
    - If Value <= Target
    - Then (Color: Red)

이 시각화는 정말 간단한 것이지만, 대시보드에서 시각화를 다루는 느낌을 익히는 데 도움이 됩니다. 시각화 박스 위에 마우스를 올리고 가장자리를 드래그하거나 클릭 버튼을 누르고 있으면 시각화의 위치와 크기를 조절할 수 있습니다.

#### 시각화 2: 콤보 차트
다음으로 또 다른 시각화 요소를 추가해 보겠습니다. 이번에는 3개월 동안의 총 판매 금액과 판매 건수 정보를 포함한 복합 차트입니다.

1. **데이터 (Data)** 탭에서 **+ SQL에서 생성 (+ Create from SQL)** 옵션을 선택하세요. </p>
2. 다음 쿼리를 입력하고 쿼리 편집 공간에서 실행하세요.

```
SELECT customer_name, 
       total_price AS Total_Sales, 
       date_format(order_date, "MM") AS Month, 
       product_category 
FROM db_catalog.db_schema.sales 
WHERE order_date >= to_date('2019-08-01')
AND order_date <=  to_date('2019-10-31');
```
</p>

3. 쿼리 이름을 **Three Month Sales**로 변경하세요. </p>
4. **제목 없는 페이지 (Untitled page)** 캔버스 탭로 돌아가기.
</p>
5. 캔버스 하단 메뉴에서 **시각화 추가 (Add a visualization)** 를 선택한 후 캔버스를 클릭해 시각화를 추가하세요.
</p>
6. 오른쪽의 **구성 패널 (Configuration Panel)** 에서 다음 설정을 선택하세요:
    - **데이터셋 (Dataset):** 3개월 판매
    - **시각화 (Visualization):** 콤보
    - **X 축:** 월
    - **Y축:**
      - **바 (Bar):** Total_Sales
        - **표시명 (Display Name):** Total Sales Value
      - **라인 (Line):** COUNT(*)
        - **표시명 (Display Name):** Count of Sales Orders
7. Y축 옆에 있는 Kebab menu 아이콘을 선택하고 **이중 축 활성화 (Enable dual axis)** 를 클릭하세요.

이중 축을 활성화하면 두 가지 다른 스케일에서 데이터 변화를 확인할 수 있습니다. 참고 바와 선 옵션이 각각 "왼쪽 Y축(바)"과 "오른쪽 Y축(선)"으로 표시됩니다. 8월과 10월의 세일즈 건수는 적었지만, 금액은 높았습니다. 9월에는 정반대입니다.

8. **왼쪽 Y축(막대) (Left Y axis (Bar))** 옆에 있는 Kebab menu 아이콘을 선택하세요.
</p>
9. **형식(Format)** 을 클릭하세요. </p>
10. **커스텀(Custom)** 을 선택하고 유형을 **Currency ($)** 로 변경하세요.

참고 축 제목이 차트 시리즈의 디스플레이 이름에 자동으로 맞게 조정됩니다. 시리즈 이름 옆의 색상 블록을 선택해 시리즈별 색상을 조절할 수도 있습니다.

#### 시각화 3: Databricks Assistant와 함께 하는 막대 차트

대시보드를 작성할 때 Databricks Assistant에 자연어 프롬프트를 제공하면 요청에 따라 차트를 자동 생성합니다. Databricks Assistant는 대시보드의 데이터 탭에 정의된 모든 데이터셋을 기반으로 차트를 구축하는 데 도움을 줄 수 있습니다. 한번 시도해 보세요.

다음 단계를 완료하세요:

1. 캔버스 하단 메뉴에서 **시각화 추가 (Add a visualization)** 를 선택한 후 캔버스를 클릭해 시각화를 추가하세요.
2. 위젯 상단의 텍스트 필드에 다음 프롬프트를 입력하세요:

    _x축에 product_category을(를) 표시하고 y축에 평균 total_sales을(를) 표시하는 막대 차트를 만드세요._

3. 응답을 생성하려면 **Enter**를 누르거나 **제출 (Submit)** 을 클릭하세요. 어시스턴트가 시각화를 제공하는 데 잠시 시간이 걸릴 수 있습니다.
4. 제공된 설명과 일치하는 막대 차트를 받게 됩니다.

  시각화가 안내와 다르거나 원하는 유형과 다르면, 필요에 따라 차트 설정을 조정할 수 있습니다.


**구성 패널로 시각화 편집**

생성된 차트를 y축에 또 다른 측정도를 추가하여 쌓인 막대 차트로 편집해 봅시다.

1. 캔버스에서 시각화 위젯을 클릭하여 설정 패널을 여세요.

2. 필요하다면 데이터세트를 **Three Months Sales**로 변경하세요.
3. 데이터 세트를 변경해야 하는 경우 Y축을 클릭한 후 열 이름을 선택하여 **Total_Sales**로 조정하십시오. 데이터세트를 변경해야 한다면 이는 매출과 3개월 매출 간 열의 이름이 다르기 때문입니다.
4. **product_category** 옆의 **–** 아이콘을 클릭하여 색상을 제거하세요. 이는 이중 축을 사용할 때 색상이 제품 카테고리가 아닌 Y-시리즈를 기준으로 설정되어야 하기 때문입니다.
5. 두 번째 **Y축**을 **Total_Sales**로 추가하세요.
6. 라벨을 클릭하고 변환을 **MIN**로 변경하세요. 
새 차트는 평균 Total_Sales과 최소 Total_Sales가 포함된 중첩 막대 차트를 보여줍니다.
7. **Y축**  오른쪽에 있는 케밥을 클릭하세요.
8. **이중 축 활성화** 옆에 체크박스를 체크하세요. 

이제 각 값이 각각의 스케일에 따라 표시되는 이중 축 막대 차트가 생깁니다. 이 시각화는 **Reagate** 카테고리가 최소 판매량이 가장 높지만, 카테고리가 평균 총 판매량이 가장 낮음을 보여줍니다.

#### 시각화 4: 산점도

주문일에 따라 세일즈 가치가 어떻게 달라지는지 이해하기 위해 산점도라는 차트를 하나 더 만들어 봅시다. 이 산점도를 만들려면 다음 단계를 수행하세요:

1. 캔버스 하단 메뉴에서 **시각화 추가 (Add a visualization)** 를 선택한 후 캔버스를 클릭해 시각화를 추가하세요. </p>
2. 오른쪽의 **구성 패널 (Configuration Panel)** 에서 다음 설정을 선택하세요: 

    - <b>데이터셋 (Dataset)</b>: 세일즈
    - <b>시각화 (Visualization)</b>: 산점도
    - <b>X 축</b>: order_date
        - **스케일 유형 (Scale Type):** 연속
        - **변환 (Transform):** 매일
    - <b>Y축</b>: total_price
        - **스케일 유형 (Scale Type):** 연속
        - **변환 (Transform):** 합

#### 텍스트 박스 추가하기

대시보드의 이름과 텍스트 설명을 위한 공간을 캔버스에 추가해 보겠습니다. 캔버스에 새 위젯을 추가하면 다른 위젯들이 자동으로 이동하여 배치 공간을 확보합니다. 마우스를 사용하여 위젯을 이동하고 크기를 조정할 수 있습니다. 위젯을 삭제하려면 해당 위젯을 선택한 후 삭제 키를 누르세요. 각 위젯의 오른쪽 상단에 있는 케밥 메뉴 아이콘을 통해 위젯을 조작할 수도 있습니다.

다음 단계를 완료하여 대시보드에 텍스트 상자를 추가하세요:

1. <p style="color: #0873FF"> <b>텍스트 상자 추가 (Add a text box)</b> 아이콘을 클릭하고 위젯을 캔버스 상단으로 드래그하세요. </p>
2. 유형: `# Retail Organization`

    **참고:** 텍스트 박스는 마크다운를 사용합니다. 포함된 텍스트의 `#` 문자는 <b>Retail 조직</b>이 1단계 제목임을 나타냅니다. 기본 마크다운 문법에 대해 더 알고 싶다면 이 <a href="https://www.markdownguide.org/basic-syntax/" target="_blank">마크다운 가이드</a>를 참고 하세요.

### 4부: 출판 및 공유

대시보드가 완성되면 게시하고 다른 사람들과 공유할 수 있습니다. 

게시된 대시보드는 워크스페이스 내 다른 사용자와 계정 단위로 등록된 사용자와 공유할 수 있습니다. 즉, Databricks 계정에 등록된 사용자는 워크스페이스 접근이나 compute 리소스가 할당되지 않았더라도 대시보드에 접근할 수 있습니다.

대시보드를 게시할 때 default 설정은 **데이터 권한 공유 (Share data permission)** 입니다. 게시된 대시보드의 데이터 공유 권한은 대시보드 시청자가 귀하의 데이터 권한을 사용하여 데이터에 접근하고 이를 지원하는 쿼리를 실행할 수 있도록 합니다. 이로 인해 공유 캐시가 가능해져 성능을 향상시킬 수 있습니다. **개별 데이터 권한 (Individual data permission)** 을 선택하면, 게시된 dashboard의 각 뷰어는 자신의 데이터 권한을 사용해 쿼리를 실행합니다. 이로 인해 갱신 운영이 더 자주 발생할 수 있습니다. 시각화 데이터의 기본 데이터 또는 대시보드를 구동하는 기본 SQL 웨어하우스에 대한 접근 권한이 없는 경우 _시각화가 표시되지 않습니다._   

대시보드가 게시되면 개인이나 소규모 그룹에게 이메일을 보내 시청자에게 알릴 수도 있습니다.

#### 대시보드 게시하기

대시보드를 게시하려면 다음 단계를 완료하세요:

1. 대시보드 오른쪽 상단의 **게시 (Publish)** 를 클릭하세요. **게시 (Publish)** 대화상자의 설정과 노트를 읽어보세요.

2. **게시 (Publish)** 대화상자에서 두 가지 데이터 권한 옵션을 볼 수 있습니다. **데이터 공유 권한 (Share data permission)** 을 선택하세요.

3. 선택적으로 email로 시청자에게 알림을 보내고 텍스트 필드에 변경 사항을 설명할 수도 있습니다.

4. 대화 오른쪽 하단의 **게시 (Publish)** 를 클릭하세요. **공유 (Sharing)** 대화 상자가 이후에 열려야 합니다. 그것이 열리지 않으면 dashboard 상단의 **게시 (Publish)** 옆에 **공유 (Share)** 를 선택할 수 있습니다.
    - 텍스트 필드를 사용해 개별 사용자를 검색하거나, dashboard를 **관리자 (Admins)** 또는 **모든 워크스페이스 사용자 (All workspace users)** 와 같은 사전 설정된 그룹과 공유할 수 있습니다. 이 창에서 **관리 가능 (Can Manage)** 또는 **편집 가능 (Can Edit)** 같은 레벨별 권한을 부여할 수 있습니다. 권한에 대한 자세한 내용은 <a href="https://docs.databricks.com/en/security/auth-authz/access-control/index.html#lakeview" target="_blank">Dashboard ACL</a>을 참조하세요.
    - **공유 (Sharing)** 대화상자의 하단에서 뷰 접근 권한을 제어합니다. 이 설정을 사용하면 모든 계정 사용자와 쉽게 공유할 수 있습니다.
3. **설정 공유 (Sharing settings)** 에서 드롭다운에서 **내 계정 내 누구나 뷰 (Anyone in my account can view)** 할 수 있습니다. 그다음 **공유 (Sharing)** 대화 상자를 닫으세요.
4. 대시보드 상단의 대시보드 이름 옆에 있는 컨텍스트 아이콘을 사용해 대시보드의 **초안 편집 (Edit Draft)** 과 **보기 출판 (View Published)** 버전 사이를 전환하세요.

**📌 참고:** 초안 대시보드를 편집할 때, 게시된 대시보드를 보는 사람들은 재게시하기 전까지 변경 사항을 볼 수 없습니다. 게시된 대시보드에는 새로운 데이터가 도착할 때마다 새로고침할 수 있는 쿼리를 기반으로 한 시각화가 포함되어 있습니다. 대시보드가 새로고침될 때 자동으로 새로운 데이터로 업데이트됩니다; 대시보드는 재게시할 필요가 없습니다.

### 5부: 인터랙티브 기능: 필드 필터
실험실에서 만든 dashboard는 보고에 적합하며, 시청자들은 이를 통해 최신 소매 판매 수치를 확인할 수 있습니다. 하지만 뷰어는 데이터를 더 탐색할 수 있는 통제 장치가 없습니다. 예를 들어, 사용자가 특정 기간의 데이터를 보고 싶을 경우, 대시보드 저자에게 연락해 변경 사항을 요청해야 합니다.

사용자가 필드나 파라미터 값을 기준으로 특정 데이터를 필터링할 수 있도록 사용자 컨트롤을 만들 수 있습니다. 필터는 대시보드 뷰어가 특정 필드를 필터링하거나 데이터세트 파라미터를 설정하여 결과를 좁힐 수 있게 해주는 위젯입니다. 

필터는 하나 이상의 데이터셋의 필드뿐만 아니라 한 번에 여러 페이지에 적용할 수도 있으며, 글로벌 필터(데이터와 대시보드 페이지 제목 사이의 대시보드 상단에 위치)를 통해 적용할 수 있습니다. 필드에 필터를 적용하면 사용자가 데이터 내 특정 값이나 값 범위에 집중할 수 있습니다. 필터는 선택한 데이터셋을 기반으로 구축된 모든 시각화에 적용됩니다.

대시보드에 필터를 추가하려면 다음 단계를 수행하세요:

1. 대시보드에서 다른 페이지로 이동했다면 다시 대시보드로 돌아가세요.


    - **대시보드 (Dashboards)** 메뉴에서 **소유자 (Owner)** 옵션을 선택하고, 뷰을 Dashboards 자신이 생성하거나 소유한 것만으로 필터링하세요. 

2. 뷰 발행 버전을 보고 있다면 대시보드의 편집 초안 버전으로 전환하세요.

3. 캔버스 하단 근처 툴바의 **필터 (Filter)** 아이콘을 클릭하세요.
4. 위젯을 대시보드 상단 근처에 배치하세요. 텍스트 박스 아래에 그것을 추가하는 게 좋을 수도 있어요. 대시보드의 위젯을 재배치하여 대시보드를 원하는 방식으로 정리할 수 있습니다.

5. 필터 위젯을 선택하면 화면 오른쪽에 필터 설정 패널이 나타납니다.

6. 다음 설정을 적용하세요:
  - **필터 (Filter)**: 단일 값
  - **필드 (Fields)**: 
      - sales.product_category 
      - 3개월간 Sales.product_category
7. 체크박스를 사용해 **제목 (Title)** 를 켜세요.
8. 위젯의 제목을 더블 클릭한 후 제목을 **제품 카테고리**로 변경하세요.
9. 필터 위젯의 드롭다운을 사용해 필터를 테스트하세요.

**📌 참고:** 필터는 필터 구성 패널에서 선택된 각 데이터세트에 적용됩니다. 선택한 모든 데이터셋은 product_category 값의 범위가 동일합니다. 대시보드 뷰어는 대시보드에서 필터링할 데이터를 선택할 때 그 목록에서 선택할 수 있습니다.

매개변수를 사용해 인터랙티브 대시보드를 만들 수도 있습니다. 매개변수를 통해 사용자는 런타임 쿼리에 값을 대입하여 데이터세트 시각화를 맞춤화할 수 있습니다. <a href="https://docs.databricks.com/en/dashboards/parameters.html" target="_blank">대시보드 매개변수는 무엇인가요?</a> 더 배우는 것은 이 과정의 범위를 벗어납니다.

그것을 수정한 후 대시보드를 다시 게시해 새 필터를 반영하도록 하세요.
