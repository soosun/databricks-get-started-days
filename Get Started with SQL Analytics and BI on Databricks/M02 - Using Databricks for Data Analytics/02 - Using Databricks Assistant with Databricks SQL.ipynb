## 02 - Databricks SQL과 함께 Databricks Assistant 사용하기

이번 수업에서는 Databricks Assistant를 활용해 쿼리 개발과 테스트를 지원하는 방법을 배우게 됩니다. 

이 수업은 `db_catalog.db_schema`에서 다음 자료들을 사용합니다:
* **sales** 테이블
* **customers** 테이블

또한 카탈로그에서 `db_schema`을(를) `db_catalog`해야 합니다.

#### 실행 설정 (필수)
아래 셀에서 **db_catalog** / **db_schema** 를 설정합니다. (**customers**, **sales** 테이블이 스키마에 있어야 합니다.)

```sql
USE CATALOG db_catalog;
USE SCHEMA db_schema;
```

```sql
USE CATALOG db_catalog;
USE SCHEMA db_schema;
```

### 1부: 글쓰기
Assistant를 사용해 다음 셀에 쿼리를 작성하십시오. 아래 셀을 선택하면 AI로 **생성 (generate)** 옵션이 제공됩니다. **생성 (generate)** 을 클릭하면 다음 프롬프트를 입력할 수 있습니다:

`"high_sales" 뷰 생성 SQL 쿼리 작성하십시오. 총 판매 가치는 10,000 이상에 대해 db_catalog.db_schema.`

실행 시 생성된 코드가 db_catalog 카탈로그에서 사용자 전용 스키마 아래에 뷰 생성되도록 하십시오.

**참고:** `db_catalog.db_schema`테이블이나 뷰를 생성할 수 없으므로, 쿼리를 수정하여 해당 뷰를 자신의 스키마에 생성해야 합니다. AI가 `db_catalog.db_schema`에 기록하면, 쓰기 권한이 없습니다. 카탈로그와 스키마가 명시되어 있지 않으면, AI가 당신의 기본 위치로 기록할 수 있습니다. AI가 반환하는 결과에 따라 차이가 있을 수 있으니, 코드가 테이블을 어디에 쓰는지 주의하세요.

프롬프트를 어떻게 수정해야 사용자 스키마에 스키마를 생성하는 쿼리를 제공할 수 있나요? 이 아래에 셀을 만들어 아이디어를 테스트하여 더 나은 프롬프트를 찾아보세요.

---
### 2부: 디버깅
다음 몇 셀에는 실행되지 않는 잘못 작성된 쿼리가 포함되어 있습니다. Assistant를 사용해 쿼리에 문제가 있는 부분을 찾아내고, 이를 적절히 다시 작성하여 실행될 수 있도록 하세요. 먼저 셀을 실행해 보세요. 문제 해결을 지원하려면 "Diagnose error"를 선택해 Databricks Assistant을 불러오세요.

```sql
SELECT * FROM sales WHERE
```

```sql
SELECT * FROM WHERE column_name = 'total_price';
```

```sql
SELECT FROM customers;
```

이 모든 간단한 예시들은 코드를 단순히 검토해도 디버깅할 수 있지만, 제대로 실행되지 않는 훨씬 더 복잡한 쿼리 사례를 상상할 수 있습니다. Databricks 어시스턴트는 문제점을 신속하게 파악하고 해결책을 제시하는 데 큰 도움이 될 수 있습니다.
