## 01 - Databricks에서 쿼리 작성

이번 수업에서는 카탈로그 탐색기에서 데이터를 발견하고 Databricks Notebooks를 사용해 데이터를 탐색하고 조작하는 쿼리를 개발하는 방법을 배우게 됩니다. 

이 수업은 카탈로그에 있는  `db_catalog` 자료를 활용합니다. 이 자료들의 사용법은 다음 지침에서 접하는 대로 설명할 것입니다.

### 1부: 데이터 발견
데이터 분석 작업을 완료하기 위해 가장 먼저 필요한 것은 요청에 맞는 적절한 데이터를 확보하는 것입니다. 적절한 데이터를 찾으려면 카탈로그 탐색기 또는 Databricks 검색을 사용해 이 프로젝트의 데이터를 찾아야 합니다. 카탈로그 탐색기를 사용하려면 아래 단계를 따라 하세요.

- 1단계: 사이드바 내비게이션에서 **카탈로그 (Catalog)** 를 선택하세요.
- 2단계: 카탈로그 선택기에서 **db_catalog** 를 찾습니다.
- 3단계: **db_schema** 스키마를 확장하세요. 이 스키마에는 세 개의 테이블이 있을 것입니다.
    - customers
    - sales
    - sales_orders

이 스키마에도 볼륨이 있다는 것을 알 수 있을 것입니다. 볼륨에는 테이블에 대응하는 CSV 파일들이 포함되어 있습니다. 볼륨에 위치한 파일에서 데이터 테이블을 만들 수는 있지만, 이는 이 내용의 범위를 벗어납니다.

이제 이 수업을 위한 데이터가 있습니다.

---

```sql
USE CATALOG db_catalog;
USE SCHEMA db_schema;
```

```sql
CREATE OR REPLACE TABLE db_catalog.db_schema.customers AS
SELECT * FROM read_files(
  '/Volumes/db_catalog/db_schema/myfiles/customers.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);

CREATE OR REPLACE TABLE db_catalog.db_schema.sales AS
SELECT * FROM read_files(
  '/Volumes/db_catalog/db_schema/myfiles/sales.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);

CREATE OR REPLACE TABLE db_catalog.db_schema.sales_orders AS
SELECT * FROM read_files(
  '/Volumes/db_catalog/db_schema/myfiles/sales_orders.csv',
  format => 'csv',
  header => true,
  inferSchema => true
);
```

### 2부: Notebooks 사용해 SQL 개발을 진행하기

여기서 쿼리를 작성하고 1부에서 확인된 데이터를 탐색하기 시작합니다.

노트북이 열린 상태에서 실행 중인 compute 클러스터에 연결되었는지 확인하십시오. 이것은 서버리스 Databricks SQL Warehouse입니다. 

SQL Warehouse는 SQL 작업 부하를 최적화하기 위해 특별히 설계되었습니다. 서버리스는 Databricks에서 성능 대비 가장 좋은 가격을 제공하지만, 규제 요건에 따라 비서버리스 compute 옵션을 사용할 경우 Databricks는 Databricks SQL Warehouse의 Pro 버전을 제공합니다. 
#### 실행 설정 (필수)
아래 셀을 실행해 **db_catalog.db_schema** 에 세 테이블을 로드합니다. (볼륨 **sqlfiles**에 CSV가 있어야 합니다.)

```sql
-- 이미 위 설정 셀에서 customers 가 로드되었습니다. 필요 시 아래만 실행해 확인합니다.
SELECT * FROM customers LIMIT 10;
```

코드를 작성하려면 노트북 칸이 필요합니다. 이 텍스트 바로 아래에 있는 **+Code** 버튼을 사용해 새 셀을 생성하세요. 이 셀과 다음 셀 사이의 공간 위에 마우스를 움직여야 **+Code** 버튼을 볼 수 있습니다.

다음 SQL 명령을 복사해서 만든 셀에 붙여넣으세요.

`SHOW TABLES IN db_schema;`

이제 그 아래에 또 다른 코드 셀을 추가하고 다음 SQL 명령어를 입력하세요.

`SELECT * FROM db_catalog.db_schema.customers;`

노트북 설정에서처럼 각 셀을 하나씩 실행할 수도 있고, 노트북 상단의 "모두 실행 (Run All)" 옵션을 선택해 노트북의 모든 셀을 순차적으로 실행할 수도 있습니다. 

**3단계 네임스페이스:** 아시다시피, 본 강의의 데이터 테이블을 참조하기 위해 우리는 세 단계 네임스페이스 구조인 `catalog.schema.table`을 사용했습니다.

---
**참고:** 이 노트북에서는 오른쪽 상단에 표시된 대로 새로운 셀이 기본적으로 SQL로 나옵니다. 이는 노트북 자체가 창 상단의 노트북 이름 옆에 표시된 대로 SQL로 설정되어 있기 때문입니다. 이 토글들은 노트북 또는 셀의 코드 유형을 변경할 수 있게 해줍니다. 예를 들어 노트북 기본값, Python 이외의 옵션을 선택하면 셀에 표시자(예: %Python 또는 %md)가 추가되어 그것이 다른 유형의 코드로 실행될 것임을 나타냅니다. SQL 창고는 SQL에만 최적화되어 있으며, 그 외에는 어떤 것도 실행하지 않습니다. 따라서 노트북이 SQL Warehouse에 연결되면 오류가 발생하지 않도록 SQL 또는 markdown 셀만 포함해야 합니다.

### 3부: 노트북에서 셀 포커스 모드를 사용하세요

이제 **포커스 모드**에 대해 이야기하겠습니다. 포커스 모드는 반쪽 화면 결과 패널로 셀을 확장할 수 있게 해줍니다. 다음 셀 오른쪽 모서리에 있는 **포커스 모드** 아이콘을 사용해 포커스 모드로 셀을 열고 댓글에 있는 세부 정보를 따라가세요.

```sql
-- 잠시 시간을 내어 포커스 모드를 탐색해 보세요. 다음과 같은 방법을 취할 수 있습니다:
-- * 셀 간 탐색
-- * 토글 어시스턴트
-- * 셀 댓글 또는 버전 이력 검토
-- * 터미널 또는 출력 창을 확장하세요.

-- 왼쪽에서 catalog explorer로 이동하세요. 이 패널을 사용해 db_catalog catalog를 찾고, 시작 시 자동으로 생성되고 이 notebook에서 설정 스크립트를 실행한 후 표시되는 본인 카탈로그도 찾을 수 있습니다.

USE CATALOG db_catalog;
USE SCHEMA db_schema;
SELECT current_catalog(), current_schema();

-- 이 기능은 현재 카탈로그와 스키마를 보여줍니다.
-- 완료 후에는 상단의 포커스 모드 종료 버튼을 선택하여 포커스 모드에서 종료하세요.
```

```sql
-- 이미 위 설정 셀에서 customers 가 로드되었습니다. 필요 시 아래만 실행해 확인합니다.
SELECT * FROM customers LIMIT 10;
```

```sql
-- 다음으로 이 코드를 실행해 스키마에서 고객 테이블을 보세요.
SELECT * FROM customers;

-- 참고로, 이 노트북에서 몇 셀 전에 설정한 기본 카탈로그와 스키마를 사용합니다.
```

결과 패널에서는 데이터가 표로 제시된 것을 볼 수 있습니다.

간단한 시각화를 만들 예정이지만, 나중에 수업에서 시각화를 더 자세히 다룰 것임을 참고해 주세요. 다음 단계를 완료하세요:

1. 쿼리 편집기의 결과 섹션에서 **+** 버튼을 클릭한 후 **시각화 (Visualization)** 를 선택하세요.
2. **시각화 유형 (Visualization Type)** 은 **바 (Bar)** 로 남겨두세요.
3. **X 열** 드롭다운을 사용해 **상태 (state)** 를 선택하세요.
4. **Y 열** 헤더 아래에서 **열 추가 (add column)** 를 클릭하고 __*__ 를 선택하세요.
5. 기본적으로 선택된 **카운트 (Count)** 를 남겨두세요.
6. 오른쪽 하단의 **저장 (Save)** 을 클릭하세요.

시각화는 결과 옆 쿼리 아래에 추가됩니다. 참고 자체 쿼리에서 그룹화나 집계를 수행할 필요가 없었지만, 시각화에는 상태별로 그룹화된 개수가 표시됩니다.

이제 시각화 제목을 클릭해 이름을 지정할 수 있고, 차트 아래의 **시각화 편집 (Edit Visualization)** 옵션을 클릭하여 시각화를 편집할 수 있습니다.
